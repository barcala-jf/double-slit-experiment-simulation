```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
```

**1. Definição das constantes da Equação**

```python
I_0 = 1.0                           # Intensidade Máxima Inicial (I_0)
y = np.linspace(-0.05, 0.05, 1000)  # Array para os pontos y (5 em 5 cm)
```

**2. Cálculo da equação com Numpy**

Para calcular a intensidade da luz $I$ em um ponto $y$ no anteparo, usamos a seguinte equação da óptica física, que combina a interferência das duas fendas e a difração da largura de cada fenda:
$$I(y) = I_0 \cos^2\left(\frac{\pi d y}{\lambda L}\right) \text{sinc}^2\left(\frac{a y}{\lambda L}\right)$$

Termo de interferência:
$$cos^2(\frac{\pi dy}{\lambda L})$$

Termo de difração:
$$sinc^2(\frac{ay}{\lambda L})$$



```python
def calcula_eq(comprimento_onda, distancia_fendas, distancia_tela, largura_fenda):
  lam = comprimento_onda * 1e-9  # nm para metros
  d = distancia_fendas * 1e-6    # micrômetros para metros
  a = largura_fenda * 1e-6       # micrômetros para metros

  # Aproximação de pequenos ângulos
  sen_theta = y / distancia_tela

  # Termo de Interferência
  fase_interferencia = (np.pi * d * sen_theta) / lam
  termo_interferencia = np.cos(fase_interferencia)**2

  # Termo difração
  arg_difracao = (a * sen_theta) / lam
  termo_difracao = np.sinc(arg_difracao)**2

  # Intensidade total
  intensidade = I_0 * termo_interferencia * termo_difracao

  return intensidade
```

**3. Estruturação em Pandas**

Para armazenar e correlacionar os dados, criamos um DataFrame em pandas

```python
def salvar_dataframe(comprimento_onda, distancia_fendas, distancia_tela, largura_fenda):
  df_resultados = pd.DataFrame({
      'Posição-y (cm)' : y * 100,
      'Intensidade-Relativa' : (calcula_eq(comprimento_onda, distancia_fendas, distancia_tela, largura_fenda))
  })

  return df_resultados
```

**4. Função para plotar o gráfico**

```python
def plot_interferencia(df_experimento):
  fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8), gridspec_kw = {'height_ratios': [2, 1]})

  # Gráfico 1D: A Curva de Intensidade
  ax1.plot(df_experimento['Posição-y (cm)'], df_experimento['Intensidade-Relativa'], color='cyan')
  ax1.set_title('Padrão de Interferência da Dupla Fenda', fontsize=16, color='white')
  ax1.set_ylabel('Intensidade Relativa', color='white')
  ax1.grid(True, alpha=0.3)
  ax1.set_facecolor('black')
  ax1.tick_params(colors='white')

  # Gráfico 2D: O Mapa de Calor simulando a projeção no anteparo
  # Repetimos a linha de intensidade verticalmente para criar uma imagem 2D
  imagem_2d = np.tile(df_experimento['Intensidade-Relativa'].values, (100, 1))

  ax2.imshow(imagem_2d, extent=[-5, 5, 0, 1], cmap='gist_heat', aspect='auto')
  ax2.set_xlabel('Posição no Anteparo (cm)', color='white')
  ax2.set_yticks([]) # Esconde o eixo Y do heatmap
  ax2.tick_params(colors='white')

  # Estética da janela geral
  fig.patch.set_facecolor('#1e1e1e')
  plt.tight_layout()
  plt.show()
```

**5. Função de Controle para vizualizar as mudanças automaticamente**

```python
def controler(comprimento_onda, distancia_fendas, distancia_tela, largura_fenda):
  df_experimento = salvar_dataframe(comprimento_onda, distancia_fendas, distancia_tela, largura_fenda)
  plot_interferencia(df_experimento)
```

**6. Vizualizar mudanças de váriaveis do sistema no gráfico**

```python
interact(controler,
         comprimento_onda = FloatSlider(value=532.8, min=300.00, max=700.00, step=10.0, description='λ (nm):'),
         distancia_fendas = FloatSlider(value=250.0, min=0.0, max=500.0, step=10.0, description='d (μm):'),
         distancia_tela =   FloatSlider(value=5.00, min=1.0, max=10.0, step=0.5, description='L (m):'),
         largura_fenda =    FloatSlider(value=10.0, min=0.0, max=1000.00, step=10.0, description='a (μm)'))
```
